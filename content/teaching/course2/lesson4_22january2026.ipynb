During an experiment a team of scientists measures the speed of thermal neutrons. Their velocities are:

velocities = [2.23, 2.01, 2.41, 2.07, 2.06, 2.31, 2.29, 2.11, 2.37, 2.25] km/s.

Using a for loop, calculate:

*   the mean velocity
*   variance
*   standard deviation
*   error on the mean

**Bonus**: the team of scientists find a way to figure out the uncertainties on each measurement. They are the following:

err_velocities = [0.01, 0.06, 0.07, 0.04, 0.02, 0.04, 0.09, 0.02, 0.05, 0.01] km/s.

Calculate the weighted mean and its uncertainty in this case.

```python
# DEFINE VARIABLES
# VAR 1
# VAR 2
# VAR 3, ..

velocities = [2.23, 2.01, 2.41, 2.07, 2.06, 2.31, 2.29, 2.11, 2.37, 2.25]
#print(velocities)

total = 0
total_sq_diff = 0
N = len(velocities)

for v in velocities:
  #print(v)

  total = total + v

print("The sum of the list is", total)

mean = total / N

print("The calculated mean is", mean)

# Now let's make another loop to calculate the variance
for v in velocities:
  total_sq_diff = total_sq_diff + (v-mean)**2

variance = total_sq_diff/N
print("The variance is", variance)

# The standard deviation is simply the square root of the variance
std_dev = variance**0.5
print("The standard deviation is", std_dev)

# Weighted mean
num = 0.0
den = 0.0
err_velocities = [0.01, 0.06, 0.07, 0.04, 0.02, 0.04, 0.09, 0.02, 0.05, 0.01]
for v, err_v in zip(velocities, err_velocities):
  print(v, '\t', err_v)

  num = num + (v/err_v**2)
  den = den + (1/err_v**2)

weighted_mean = num/den
err_weighted_mean = (1.0/den)**0.5
print("The weighted mean is", weighted_mean)
print("The error on the weighted mean is", err_weighted_mean)
```
