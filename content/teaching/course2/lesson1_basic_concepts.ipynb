##Containers

Python includes several built-in container types: lists, dictionaries, sets, and tuples.

We will see tuple and lists. The main difference is that tuples are immutable as opposed to lists which are mutable. You can't change a tuple. You can change a list.



```python
my_tuple = (1, 2)
my_list = [1, 2]

print(type(my_tuple))
print(type(my_list))

print(my_tuple)
print(my_list)
```

**NOTE: Unless you are very sure, use lists instead of tuples, since they allow you to modify/add data.**

```python
my_list = ['Niccolo', 'Lana']
print(my_list)
my_list.append('Frank')
print(my_list)

my_name = my_list[0]
print(my_name)
print(type(my_name))
```
