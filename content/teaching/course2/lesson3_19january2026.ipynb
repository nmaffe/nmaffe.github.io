## Slicing
In addition to accessing list elements one at a time, Python provides concise syntax to access sublists; this is known as slicing:



19 January, 2026

In general, if you have a list x[index] returns the value associated to index.
Remember that python is zero-indexed, meaning the first index is 0.

```python
x = [0, 1, 2, 3, 4]
print(type(x))
print(x)
```

```python
#print(x[1])
#print(x[4])
#print(x[2:4])      # Get a slice from index 2 to 4 (exclusive); prints "[2, 3]"
#print(x[2:])       # Get a slice from index 2 to the end; prints "[2, 3, 4]"
#print(x[:2])       # Get a slice from the start to index 2 (exclusive); prints "[0, 1]"
#print(x[:])        # Get a slice of the whole list; prints ["0, 1, 2, 3, 4]"
#print(x[:-1])      # Slice indices can be negative; prints ["0, 1, 2, 3]"
x[2:4] = [8, 9]    # Assign a new sublist to a slice
print(x)           # Prints "[0, 1, 8, 9, 4]"
```

```python
list_names = ['Niccolo', 'Lena', 'Artyom']
print(list_names)

list_names[0] = 'Sarah'
print(list_names)

print(list_names[2])
```

## For loop
A `for` loop is used for iterating over a sequence, for example a tuple or a list.

```python
fruits = ["apple", "banana", "cherry"]
for i in fruits:
  print(i)
```

```python
for x in [0, 2, 4, 6, 8, 10]:
  print(f"At this iteration x = {x}")
  double_x = 2*x
  print(f"The double of this number is {double_x}")

print('Finished')
```
