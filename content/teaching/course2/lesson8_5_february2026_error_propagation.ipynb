```python
import math
```

```python
V = 12.0        # volts
sigma_V = 0.2   # volts
I = 1.5         # ampere
sigma_I = 0.05  # ampere
```

```python
P = V * I # watts
print('The power is ', P, 'watts.')

sigma_P = P * math.sqrt((sigma_V/V)**2 + (sigma_I/I)**2)

print('The uncertainty in the power is ', sigma_P)
```

```python
def mult_and_error(x, sigma_x, y, sigma_y):

  z = x * y
  sigma_z = z * math.sqrt((sigma_x/x)**2 + (sigma_y/y)**2)

  return z, sigma_z

def divide_and_error(x, sigma_x, y, sigma_y):

  z = x / y
  sigma_z = z * math.sqrt((sigma_x/x)**2 + (sigma_y/y)**2)

  return z, sigma_z
```

```python
power, error_power = mult_and_error(V, sigma_V, I, sigma_I)
print('Power: ', power, 'Uncertainty: ', error_power)
```

```python
area, error_area = mult_and_error(30.0, 0.1, 20.0, 0.5)
print('Area: ', area, 'Uncertainty: ', error_area)
```
