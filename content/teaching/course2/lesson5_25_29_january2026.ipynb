# Fuctions
A function is a block of code which only runs when it is called.

You can pass data, known as parameters, into a function.

A function can return data as a result.


A function is defined by ```def```. The parameters have to be specified in the (). The ```return``` keywords specifies that this will be returned by the function.



````python
def multiply_by_five_and_add_three(x):
  return 5 * x + 3

def double_x_plus_y(x,y):
  return 2 * x + y

pippo = multiply_by_five_and_add_three(2)
print('pippo is', pippo)


franco = double_x_plus_y(4, 3)
print('franco is', franco)
````

````python
def sign(x):
    if x > 0:
        return 'positive'
    elif x < 0:
        return 'negative'
    else:
        return 'zero'

for x in [-1, 0, 1]:
  res = sign(x)
  print(res)
````

````python
def Binomial(k, p, n):
  return (p**k)*(1-p)**(n-k)*math.factorial(n)/(math.factorial(k)*math.factorial(n-k))

def Poisson(l, k):
  return (l**k)*math.exp(-l)/math.factorial(k)

def Gaussian(x, mu, sigma):
  pi = math.pi
  return (1/math.sqrt(sigma*(2*pi)))*np.exp(-(x-mu)**2/(2*sigma**2))
````
