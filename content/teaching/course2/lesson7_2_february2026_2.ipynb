```python
# math and numpy are popular libraries with a lot (A LOT) of pre-defined functions
import math
import numpy as np
```

```python
def Binomial(k, p, n):
  return (p**k)*(1-p)**(n-k)*math.factorial(n)/(math.factorial(k)*math.factorial(n-k))

def Poisson(l, k):
  return (l**k)*math.exp(-l)/math.factorial(k)

def Gaussian(x, mu, sigma):
  pi = math.pi
  return (1/math.sqrt(sigma*(2*pi)))*np.exp(-(x-mu)**2/(2*sigma**2))
```

```python
# EX3
lam = 93/100
print('Prob exactly 3 earthquakes:', Poisson(l=lam, k=3))

prob_tot = 0
for k in range(4):
  #print(k)
  prob_tot = prob_tot + Poisson(l=lam, k=k)
print('Prob less than 4 is:', prob_tot)
```
